```python
import numpy as np

file = open("../../Downloads/glove.6B/glove.6B.50d.txt", "r")

text = file.read()

file.close()

lines = text.split('\n')


words = []
vectors = []

for line in lines:
    if not line.strip():
        continue
        
    embedding = line.split(' ')
    words.append(embedding[0])
    vectors.append(np.array(embedding[1:], dtype=float))


vectors = np.array(vectors)

normalized_vectors = vectors / np.linalg.norm(
    vectors, axis=1, keepdims=True
)
```

```python
idx_man = words.index("man")
idx_woman = words.index("woman")

v1 = vectors[idx_woman]
v2 = vectors[idx_man]

v3 = v1 - v2

vectors2 = vectors + v3

vectors2_norm = vectors2 / np.linalg.norm(vectors2, axis=1, keepdims=True)

batch_size = 1000

best_matches = []

for i in range(0, len(vectors2_norm), batch_size):
    batch = vectors2_norm[i:i + batch_size]

    similarities = batch @ normalized_vectors.T

    best_matches.extend(np.argmax(similarities, axis=1))
    
    if i % 10000 == 0:
        print(f"Progress: {i}/{len(vectors2_norm)}")

best_matches = np.array(best_matches)

for i, best_match in enumerate(best_matches):
    if best_match != i:
        print(words[i] + " " + words[best_match])
```
