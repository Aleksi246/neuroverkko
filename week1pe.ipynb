```python
import pandas as pd
import numpy as np
import keras
from keras import layers
from keras.losses import CategoricalCrossentropy
from keras.optimizers import Adam

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

df = pd.read_csv(url, header=None)
matrix = np.array(df.values.tolist())

np.random.shuffle(matrix)

onehot = []
for i in range(len(df)):
    if matrix[i][4] == "Iris-setosa":
        onehot.append([1,0,0])
    if matrix[i][4] == "Iris-versicolor":
        onehot.append([0,1,0])
    if matrix[i][4] == "Iris-virginica":
        onehot.append ([0,0,1])

onehot = np.asarray(onehot, dtype=np.float32)
#poistaa viimeisen kolumnin
matrix = matrix[:, :-1]

#stringit numeroiksi
matrix = matrix.astype(float)

#scaalataan data
X_mean = matrix.mean(axis=0)
X_std = matrix.std(axis=0)

matrix_scaled = (matrix - X_mean) / X_std
```

4 inputs hidden layer yhdellä neuronilla (jännä että silti toimii tein sen vain satunnaisena kokeiluna kun kiinnosti) 3 outputs. learning rate on 0.006 ja loss lasketaan categorical entropy.Batch size pienentäminen paransi mallia.

```python
inputs = keras.Input(shape=(4,))
x = layers.Dense(1, activation="relu")(inputs)
outputs = layers.Dense(3, activation="softmax")(x)

model = keras.Model(inputs=inputs, outputs=outputs)

model.compile(
    optimizer=Adam(learning_rate=0.006),
    loss='categorical_crossentropy',
    metrics=["accuracy"]
)

history = model.fit(matrix_scaled, onehot, epochs=100, batch_size=2,validation_split=0.2)
```
