ensimmäisessä koodi solussa luen tekstin ja jaan sen tupleisi missä 0 on sana ja 1 on sen selittävä vektori.


toisessa solussa on functio joka ottaa kolme sanaa, löytää näiden sanojen vektorin ja laskee yhtälön vec1 - vec2 + vec3 tulokesen.
tämä tulos tallennetaan uudessa vektorissa ja verrataan kaikkii sanaston vektoreihin cosini samankaltaisuuden perusteella. lopuksi tulostetaan samankaltaisimman vektorin sana.


teoriassa esimerkiksi vektorin "woman" ja vektorin "man" ero olisi samankaltainen kuin vektorin "girl" ja vektorin "boy" ero  
: woman - man = girl - boy  
tämä saataisiin muokattua muotoon: woman - man + boy = girl 

tehtävän esimerkki tapaus: woman - man + king antaa tuloksen king koska king vektori on silti lähempänä uutta vektoria kuin queen.
sain tulokseksi queen kun kerroin vektorin (woman - man) skalaarilla 1.2 eli 1.2(woman-man) + king = queen

```python
import numpy as np

file = open("../../downloads/glove.6B.50d.txt", "r")
#teksti tiedostossa on tyyppiä  "sana 0.14 0.25241 1.4224 ... 0.013 newline" float arvoja on 300

text = file.read()

file.close()

lines = text.split('\n')

split_lines = []

for line in lines:
    if not line.strip():
        continue
        
    embedding = line.split(' ')
    split_lines.append((embedding[0],np.array(embedding[1:], dtype=float)))
```

```python

def fun(word1, word2, word3):
    #word1 - word2 = word4 - word3
    #example: woman - man = girl - boy
    #woman - man + boy = girl
    v1 = None
    v2 = None
    v3 = None
    for vector in split_lines:
        if vector[0] == word1:
            v1 = vector[1] 
        if vector[0] == word2:
            v2 = vector[1]
        if vector[0] == word3:
            v3 = vector[1]

    if v1 is None:
        print(word1 + " not found")
        return
    if v2 is None:
        print(word2 + " not found")
        return
    if v3 is None:
        print(word3 + " not found")
        return
    
    v4 = v1 - v2 + v3
    
    cosine_sim_limit = -1
    v5 = None
    
    for vector in split_lines:
        cos_sim = np.dot(v4, vector[1]) / (np.linalg.norm(v4) * np.linalg.norm(vector[1]))
        if cos_sim > cosine_sim_limit:
            cosine_sim_limit = cos_sim
            v5 = vector
    
    print(v5[0])
```

```python
fun("woman","man","boy")
#woman - man = girl - boy
```

```python
fun("hitler","germany","italy")
#hitler - germany = mussolini - italy
```

```python
fun("woman","man","king")
#woman - man = queen - king
```

```python
fun("finland","helsinki","stockholm")
#finland - helsinki = sweden - stockholm
```
