```python
import numpy as np
import matplotlib.pyplot as plt

#tehtava 1
def f(x):
    y = x**4 + 5*x**3 -7*x
    return y

def df(x):
    y = 4*x**3 + 15*x**2 -7
    return y

x_values = np.linspace(-5,2,1000)
y_values = f(x_values)
plt.plot(x_values, y_values)
plt.grid()
plt.show()
```

```python
x = 0
path_x = np.array([x])
alpha = 0.01

x = x - alpha*df(x)
path_x = np.append(path_x, x)

while path_x[-1]!= path_x[-2]:
    x = x - alpha*df(x)
    path_x = np.append(path_x, x)

print(path_x[-1], len(path_x))

path_y = f(path_x)
plt.plot(path_x, path_y, '.')

xx = np.linspace(-5,2,1000)
y_values = f(xx)
plt.plot(xx, y_values)
plt.grid()
plt.show()
```

```python
#tehtava 2

def function(x,y):
    z = 3*x**2 + 3*x*y + 2*y**2 +3*x - 5*y
    return z
    
def dfunction(x,y):
    dx = 6*x + 3*y + 3
    dy = 3*x + 4*y - 5
    return dx, dy

x_values = np.linspace(-20,20,1000)
y_values = np.linspace(-20,20,1000)
(X, Y) = np.meshgrid(x_values, y_values)
Z = function(X,Y)

plt.contour(X, Y, Z)

plt.grid()
plt.show()
```

```python
x = -15
y = -15
alpha = 0.01

path_x = np.array([x])
path_y = np.array([y])

dx, dy = dfunction(x, y)
x = x - alpha*dx
y = y - alpha*dy
path_x = np.append(path_x, x)
path_y = np.append(path_y, y)

while path_x[-1]!= path_x[-2] or path_y[-1]!= path_y[-2]:
    dx, dy = dfunction(x, y)
    x = x - alpha*dx
    y = y - alpha*dy
    path_x = np.append(path_x, x)
    path_y = np.append(path_y, y)

print("x = ", x, "     y = ", y, " steps: ", len(path_x))

x_values = np.linspace(-20,20,1000)
y_values = np.linspace(-20,20,1000)
(X, Y) = np.meshgrid(x_values, y_values)
Z = function(X,Y)
plt.contour(X, Y, Z)
plt.plot(path_x, path_y, 'r+')
plt.axis('equal')
plt.show()
```
