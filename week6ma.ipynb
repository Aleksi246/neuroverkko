```python
import numpy as np
import math

def y_pred(theta_0,theta_1,x):
    y_hat = 1/(1+math.exp(-(theta_0 + theta_1*x)))

    return y_hat


x = [1.1,4.1,2.7,3.9,7.1,1.3,9.7,0.7,3.4,6.0]
y = [0,1,0,0,1,0,1,0,1,0]

theta_0 = 0
theta_1 = 0
alpha = 0.01

z = 0
while z < 1000000:
    z += 1
    
    y_hats = []
    for i in x:
        yhat = y_pred(theta_0,theta_1,i)
        y_hats.append(yhat)

    d_t_0 = 0
    for i in range(len(y)):
        d_t_0 += y_hats[i] - y[i]

    d_t_1 = 0
    for i in range(len(y)):
        d_t_1 += (y_hats[i] - y[i])*x[i]

    theta_0 = theta_0 - alpha*d_t_0

    theta_1 = theta_1 - alpha*d_t_1

print(theta_0,theta_1)
```
